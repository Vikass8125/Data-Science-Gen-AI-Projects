```python
from langsmith import Client
from langchain.agents import create_agent
```

```python
# Initialize the client
client = Client()

# Pull a prompt from the Hub
prompt = client.pull_prompt("hwchase17/react")
```

```python
# Convert to string if you need the actual text
prompt_string = prompt.template  # Get the template string
agent = create_agent(
    model="gpt-3.5-turbo",
    tools=[search_tool],
    system_prompt=prompt_string
)
```

```python
response = agent.invoke({"messages": "What is the capital of France?"})
```
